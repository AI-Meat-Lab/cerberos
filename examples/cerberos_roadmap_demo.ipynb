# Cerberos roadmap demo
This notebook demonstrates strict clustering, multi-metric verification, whole-cluster balancing, and boundary diagnostics.

```python
from cerberos_peptide_splitter import RunConfig, boundary_records
from cerberos_peptide_splitter.fasta import parse_fasta
from cerberos_peptide_splitter.split import split_records

records = parse_fasta('peptides.fasta')
config = RunConfig(
    strict_clustering=True,
    verification_metrics=['containment', 'cosine'],
    metric_aggregation='mean',
    kmer_sizes=[2, 3, 4],
    balance=True,
    balance_metric='wasserstein',
    local_search_iterations=5,
)
splits = split_records(records, config)
boundary = boundary_records(splits, k=3, threshold=config.similarity_threshold)
print({name: len(values) for name, values in splits.items()})
print('boundary candidates:', len(boundary))
```

Inspect `config.last_clustering_stats`, `config.last_assignment_stats`, `stats.json`, and `report.html` before using a split.
